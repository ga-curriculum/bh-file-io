## Part I: File IO

```python
# Open a connection to a file with a "write" (w) flag
file = open('grocery.txt', 'w')
```

```python
# You can write individual bits of text to a file.
# Check to see if it worked!
file.write('coffee\n')
file.write('bread\n')
```

```python
# Actually, nothing really gets written until you close teh connection.
file.close()
```

```python
# Connecting to the same file overwrites the old file unless you
# use an "append" (a) flag.
file = open('grocery.txt', 'a')
file.writelines(['turkey\n', 'cheese\n'])
file.close()
```

```python
# .readlines() is the most common way to read a file.
file = open('grocery.txt', 'r')
lines = file.readlines()
file.close()
```

```python
lines
```

```python
# Context manager
# But pepole never do this - they mostly just use a "with" context manager
# to automatically open and close files. This also includes protection from
# your code bugging out.
with open('todo.txt', 'w') as f:
    f.writelines([
        'feed cat\n',
        'finish lab\n',
        'run a mile\n'
    ])
```

```python
with open('todo.txt', 'r') as f:
    lines = f.readlines()
```

```python
lines
```

## Part II: JSON IO

```python
import json
```

```python
data = {
    'movie': 'Shrek 2',
    'music': ['Nickelback', '311'],
    'book': 'The Da Vinci Code'
}
```

```python
# Writing and reading json files is the same as for regular text files,
# but you need to use json.dump() and json.load(), respectively.
with open('favorites.json', 'w') as f:
    json.dump(data, f)
```

```python
with open('favorites.json', 'r') as f:
    data_in = json.load(f)
```

```python
data_in
```

# Part III: Pickling

Pickling is a way to save and load entire model objects! This is extremely useful when

1. Your models take a long time to fit
1. You want to serve predictions from your model somewhere else (maybe in a streamlit app!)

```python
import numpy as np
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
import pickle
```

```python
# Make fake data
X, y = make_regression(1000, 10)
```

```python
X.shape
```

```python
y.shape
```

```python
m = LinearRegression()
m.fit(X, y)
```

```python
m.coef_
```

```python
# Pickling a model requires the special "b" flag.
# sklearn models are stored as bytecode, which is not human readable,
# so the b flag tells Python we'll be saving data to a text file that
# is not a typical "flat" file.
with open('my-model.pkl', 'wb') as f:
    pickle.dump(m, f)
```

```python
with open('my-model.pkl', 'rb') as f:
    m_in = pickle.load(f)
```

```python
m_in.coef_
```
