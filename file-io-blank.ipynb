# Part III: Pickling

Pickling is a way to save and load entire model objects! This is extremely useful when

1. Your models take a long time to fit
1. You want to serve predictions from your model somewhere else (maybe in a streamlit app!)

```python
import numpy as np
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
import pickle
```

```python
# Make fake data
X, y = make_regression(1000, 10)
```

```python
X.shape
```

```python
y.shape
```

```python
m = LinearRegression()
m.fit(X, y)
m.coef_
```

```python
# Pickling a model requires the special "b" flag.
# sklearn models are stored as bytecode, which is not human readable,
# so the b flag tells Python we'll be saving data to a text file that
# is not a typical "flat" file.
```
